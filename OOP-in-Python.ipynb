## Assignment 1: Student Class (Encapsulation)

```python
class Student:
    def __init__(self, student_id, name, program, gpa):
        self.student_id = student_id
        self.name = name
        self.program = program
        self.__gpa = None
        self.set_gpa(gpa)

    def get_gpa(self):
        return self.__gpa

    def set_gpa(self, gpa):
        if 0.0 <= gpa <= 4.0:
            self.__gpa = gpa
        else:
            raise ValueError("GPA must be between 0.0 and 4.0")

    def display_info(self):
        print(f"ID: {self.student_id}, Name: {self.name}, Program: {self.program}, GPA: {self.__gpa}")

# Test
s1 = Student("S001", "Abel", "CS", 3.5)
s1.display_info()
```

## Assignment 2: Animal Classes (Inheritance)

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cow(Animal):
    def speak(self):
        return "Moo!"

class Bird(Animal):
    def speak(self):
        return "Chirp!"

animals = [Dog("Rex", 5), Cow("Bessie", 3), Bird("Tweety", 1)]
for a in animals:
    print(f"{a.name} says: {a.speak()}")
```

## Assignment 3: Bioinformatics Sequence System

```python
class Sequence:
    def __init__(self, seq_id, sequence):
        self.seq_id = seq_id
        self.sequence = sequence.upper()

    def length(self):
        return len(self.sequence)

class DNASequence(Sequence):
    def gc_content(self):
        gc = self.sequence.count("G") + self.sequence.count("C")
        return gc / len(self.sequence) * 100

class RNASequence(Sequence):
    def to_dna(self):
        return self.sequence.replace("U", "T")

class ProteinSequence(Sequence):
    def count_amino_acids(self):
        return {aa: self.sequence.count(aa) for aa in set(self.sequence)}

dna = DNASequence("DNA1", "ATGCGTACGTAGCTAGCTAGCTA")
print("Length:", dna.length(), "GC%:", dna.gc_content())
```

## Assignment 4: Library Management System

```python
class Book:
    def __init__(self, title):
        self.title = title
        self.available = True

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed = []

class Library:
    def __init__(self):
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)

    def register_member(self, member):
        self.members.append(member)

    def borrow_book(self, member, book):
        if book.available:
            book.available = False
            member.borrowed.append(book)
        else:
            print("Book not available")

    def return_book(self, member, book):
        if book in member.borrowed:
            book.available = True
            member.borrowed.remove(book)

lib = Library()
b1 = Book("Python Basics")
m1 = Member("Abel")
lib.add_book(b1)
lib.register_member(m1)
lib.borrow_book(m1, b1)
```

## Assignment 5: Abstract Shapes

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self): pass
    @abstractmethod
    def perimeter(self): pass

class Circle(Shape):
    def __init__(self, r): self.r = r
    def area(self): return math.pi * self.r**2
    def perimeter(self): return 2 * math.pi * self.r

class Rectangle(Shape):
    def __init__(self, l, w): self.l, self.w = l, w
    def area(self): return self.l * self.w
    def perimeter(self): return 2*(self.l+self.w)

class Triangle(Shape):
    def __init__(self, a, b, c): self.a, self.b, self.c = a, b, c
    def area(self): 
        s = (self.a+self.b+self.c)/2
        return math.sqrt(s*(s-self.a)*(s-self.b)*(s-self.c))
    def perimeter(self): return self.a+self.b+self.c

shapes = [Circle(3), Rectangle(4,5), Triangle(3,4,5)]
for s in shapes:
    print("Area:", s.area(), "Perimeter:", s.perimeter())
```

## Assignment 6: Variant Annotation

```python
class Variant:
    def __init__(self, chrom, pos, ref, alt):
        self.chrom, self.pos, self.ref, self.alt = chrom, pos, ref, alt

    def length(self):
        return abs(len(self.alt) - len(self.ref))

    def summary(self):
        return f"{self.chrom}:{self.pos} {self.ref}>{self.alt}"

class SNP(Variant):
    def type(self): return "SNP"

class Insertion(Variant):
    def type(self): return "Insertion"

class Deletion(Variant):
    def type(self): return "Deletion"

v = SNP("chr1", 12345, "A", "G")
print(v.type(), v.summary(), "Length:", v.length())
```

## Assignment 7: Alternative Constructors

```python
class Genome:
    def __init__(self, name, species, size):
        self.name, self.species, self.size = name, species, size

    @classmethod
    def from_csv_line(cls, line):
        name, species, size = line.split(",")
        return cls(name, species, int(size))

    @classmethod
    def from_dict(cls, d):
        return cls(d["name"], d["species"], d["size"])

g1 = Genome.from_csv_line("NDama,Bos taurus,2800")
print(g1.name, g1.species, g1.size)
```

## Assignment 8: Static vs Class Methods

```python
class DNAUtils:
    valid_bases = "ATGC"

    @staticmethod
    def is_valid_dna(seq):
        return all(base in DNAUtils.valid_bases for base in seq)

    @staticmethod
    def gc_content(seq):
        gc = seq.count("G") + seq.count("C")
        return gc / len(seq) * 100

    @classmethod
    def update_valid_bases(cls, new_bases):
        cls.valid_bases = new_bases

print(DNAUtils.is_valid_dna("ATGC"))
print(DNAUtils.gc_content("ATGCATGC"))
```

## Assignment 9: Operator Overloading

```python
class Sequence:
    def __init__(self, seq):
        self.seq = seq

    def __len__(self):
        return len(self.seq)

    def __str__(self):
        return f"{self.seq}"

    def __eq__(self, other):
        return self.seq == other.seq

    def __add__(self, other):
        return Sequence(self.seq + other.seq)

s1 = Sequence("ATGC")
s2 = Sequence("GCTA")
print(s1.__len__())
print(s1.__str__())
print(s1.__eq__(s2))
print(s1.__add__(s2))
```
